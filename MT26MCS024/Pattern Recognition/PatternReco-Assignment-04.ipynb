```python
# importing Libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

print("Debug:01:: Libraries imported successfully")

filepath="Dataset/heart.csv"
```

```python
from numpy._core import fromnumeric
class PatternClassifiers :

    def __init__(self) :
        self.classes = None
        self.features = None
        self.priors = {}
        self.parameters={}

    def displayParam(self):
        print(self.parameters)


    def bayesianClassifier(self,x, mean, variance) :
        if (variance == 0 ) :
            variance=0.00000001

        exponent= np.exp((-(x-mean)**2) / (variance))
        probability=exponent / np.sqrt(2 * 3.14179* variance)

        return probability


    def fit(self, featureDF, classifications) :
        self.features=list(featureDF.columns)
        self.classes=list(set(classifications))

        total_records = len(classifications)

        for c in self.classes :
            X_c=featureDF[classifications==c]
            self.priors[c]=len(X_c) / total_records

            self.parameters[c] = {}

            for f in self.features :
                mean_val=X_c[f].mean()

                variance_val=X_c[f].var(ddof=2)

                self.parameters[c][f]={"mean":mean_val,"variance":variance_val}

    def predict(self, X):
        predictions = []
        
        # Using iterrows() to properly extract rows from a Pandas DataFrame
        for index, row in X.iterrows():
            # Call predict_single(row) and store it
            prediction = self.predict_single(row)
            predictions.append(prediction)
                
        return predictions

    def predict_single(self, row):
        probabilities = {}
        
        for c in self.classes:

            probability = self.priors[c]
            
            for f in self.features:
                mean = self.parameters[c][f]['mean']
                variance = self.parameters[c][f]['variance']


                likelihood = self.bayesianClassifier(row[f], mean, variance)
                
                probability = probability * likelihood
                

            probabilities[c] = probability


        predicted_class = max(probabilities, key=probabilities.get)

        return predicted_class

    
```

```python
# importing dataset
dataset = pd.read_csv(filepath)

dataset.head(20)


```

```python
# Class Extractions and Data Spliting

feature_set=dataset.drop('target',axis=1)
class_set=dataset["target"].map({0:"Disease-free",1:"Diseased"})
dataset.head(5)

print(feature_set)
print(class_set)


feature_set_train, feature_set_test, class_set_train, class_set_test= train_test_split(
    feature_set,
    class_set,
    test_size=0.2,
    random_state=12,
    stratify=class_set
)
```

```python
classifier = PatternClassifiers()

classifier.fit(feature_set_train,class_set_train)
classifier.displayParam()
```

```python
class_predict = classifier.predict(feature_set_test)

print(class_predict)
```

```python
correct_predictions = sum(1 for true, pred in zip(class_set_test, class_predict) if true == pred)


total_predictions = len(class_set_test)
accuracy = correct_predictions / total_predictions

print(f"Accuracy: {accuracy * 100:.2f}%\n")
```

```python
# Confusion Matrix

print("--- Confusion Matrix ---")
conf_matrix = confusion_matrix(class_set_test, class_predict)
print(conf_matrix, "\n")
```

```python
# Classification Report

print("--- Classification Report ---")
class_report = classification_report(class_set_test, class_predict)
print(class_report)
```
